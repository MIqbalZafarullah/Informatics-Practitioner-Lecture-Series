# Perancangan Dashboard Penjualan Berpusat pada Pengguna

## 1. Latar Belakang

Dalam era ledakan data, keberhasilan suatu sistem informasi tidak semata-mata diukur dari kelengkapan fitur teknis atau kompleksitas basis data di baliknya, melainkan dari bagaimana informasi tersebut disajikan kepada pengguna manusia. Pendekatan *Human-Computer Interaction* (HCI) dan *User-Centered Design* (UCD) menekankan bahwa sistem harus beradaptasi dengan cara manusia memproses rangsangan visual dan kognitif, bukan sebaliknya.

Pepatah penting dalam HCI adalah *"You are not your user"*. Pengembang sering kali tergoda membuat dashboard dengan beragam warna mencolok, efek dekoratif, atau susunan metrik yang padat tanpa mempertimbangkan beban kerja mental (*cognitive load*) pengguna akhir. Dashboard yang buruk sering kali memicu kelelahan visual (*visual fatigue*), salah tafsir skala angka, dan lambatnya pengambilan keputusan operasional.

Studi kasus ini merancang dan membandingkan dua versi antarmuka dashboard penjualan:
1. **Versi Awal (Sebelum Perbaikan)**: Mewakili desain yang secara teknis menampilkan data yang benar, namun mengabaikan prinsip ergonomi visual (minim hierarki, kontras buruk, ketiadaan kartu ringkasan KPI, dan beban mengingat informasi tinggi).
2. **Versi Perbaikan (User-Centered)**: Menerapkan prinsip Gestalt, tata letak berbasis hierarki visual, kartu KPI ringkas, skala grafis yang jujur, serta filter interaktif menggunakan widget berbasis Python.

## 2. Tujuan Studi Kasus

1. Mengidentifikasi kebutuhan pengguna toko ritel melalui profil persona fiktif yang realistis beserta perumusan 3 kebutuhan fungsional dan 3 nonfungsional.
2. Membangkitkan data sintetis penjualan bulanan selama 12 bulan untuk 3 kategori produk dengan seed terkontrol (`seed = 42`).
3. Membangun visualisasi komparatif "Sebelum" vs "Sesudah" perbaikan antarmuka untuk menunjukkan dampak penerapan prinsip HCI secara nyata.
4. Mengimplementasikan filter interaktif berbasis `ipywidgets` (dropdown kategori) yang berfungsi langsung di lingkungan Google Colab.
5. Melakukan evaluasi heuristik mandiri berbasis 5 prinsip usability Nielsen/Norman serta menyusun 3 skenario tugas uji coba untuk pengujian pengguna (*usability testing*) di masa depan.

- **Kendala Utama (*Frustrations*)**: Sering pusing melihat laporan spreadsheet yang penuh baris angka dan tidak memiliki waktu luang untuk membaca manual penggunaan aplikasi yang panjang.

### Kebutuhan Fungsional (Functional Requirements - FR)
1. **FR-1 (Ringkasan Metrik Utama)**: Sistem harus menampilkan ringkasan metrik performa utama (*KPI Cards*): Total Penjualan (Omzet dlm Rupiah), Total Transaksi (Volume), dan Rata-rata Nilai Belanja per Transaksi (*Average Order Value* / AOV).
2. **FR-2 (Visualisasi Tren Temporal)**: Sistem harus menyajikan visualisasi tren penjualan bulanan selama periode 12 bulan dalam bentuk grafik garis (*line chart*) yang komunikatif.
3. **FR-3 (Penyaringan Data Interaktif)**: Sistem harus menyediakan elemen kendali filter (*dropdown*) berdasarkan kategori produk sehingga seluruh grafik dan kartu metrik dapat diperbarui seketika.

### Kebutuhan Nonfungsional (Non-Functional Requirements - NFR)
1. **NFR-1 (Clarity & Readability)**: Teks dan label wajib menggunakan font sans-serif dengan kontras tinggi terhadap latar belakang, serta angka diformat dengan pemisah ribuan dan satuan mata uang Rupiah (Rp) yang konsisten.
2. **NFR-2 (Performance & Responsiveness)**: Pembaruan grafik saat filter dropdown diubah harus berlangsung cepat ($< 1$ detik) di Google Colab tanpa membutuhkan *refresh* halaman secara manual.
3. **NFR-3 (Recognition over Recall)**: Pilihan filter yang aktif serta satuan sumbu harus selalu terlihat secara eksplisit di layar sehingga pengguna tidak perlu mengingat status filter yang sedang berjalan.

## 4. Data dan Spesifikasi Simulasi

Data penjualan dibangkitkan secara sintetis dengan `numpy.random.RandomState(seed=42)`:
- **Periode**: 12 Bulan (Januari hingga Desember).
- **Kategori Produk**: 3 kategori:
  1. *Makanan & Minuman* (FMCG)
  2. *Kebutuhan Rumah Tangga* (Cleaning & Toiletries)
  3. *Alat Tulis & Kantor* (Stationery)
- **Variabel Data**:
  - `bulan`: 1 sampai 12
  - `nama_bulan`: Jan, Feb, Mar, ..., Des
  - `kategori`: Kategori produk
  - `jumlah_transaksi`: Jumlah nota transaksi (unit)
  - `total_penjualan`: Total nominal penjualan (Rupiah)
  - `aov`: Nilai rata-rata transaksi (`total_penjualan / jumlah_transaksi`)

## 5. Metode Evaluasi Desain Antarmuka

Evaluasi antarmuka pada studi kasus ini dilakukan melalui **Evaluasi Heuristik Mandiri (Expert Review / Self-Heuristic)** yang berpedoman pada 5 prinsip utama Jakob Nielsen:
1. **Visibility of System Status**: Pengguna selalu mendapatkan informasi yang jelas mengenai data kategori apa yang sedang ditampilkan.
2. **Consistency and Standards**: Penamaan istilah, format mata uang (Rp), dan palet warna konsisten di seluruh elemen visual.
3. **Recognition Rather than Recall**: Elemen kunci dan opsi pilihan selalu terpampang nyata, bukan tersembunyi di balik menu berlapis.
4. **Aesthetic and Minimalist Design**: Menghilangkan dekorasi berlebihan (*chart junk*), grid yang terlalu tebal, atau kombinasi warna kontras menyakitkan mata.
5. **Feedback**: Perubahan filter seketika memperbarui angka ringkasan dan judul grafik untuk memberikan konfirmasi aksi.

## 6. Implementasi Python

### 6.1 Import Library dan Bangkitkan Data Sintetis

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import ipywidgets as widgets
from IPython.display import display, clear_output
import os

SEED = 42
rng = np.random.RandomState(SEED)

# Daftar bulan dan kategori
bulan_list = ['Jan', 'Feb', 'Mar', 'Apr', 'Mei', 'Jun', 'Jul', 'Agu', 'Sep', 'Okt', 'Nov', 'Des']
kategori_list = ['Makanan & Minuman', 'Kebutuhan Rumah Tangga', 'Alat Tulis & Kantor']

# Bangkitkan data sintetis 12 bulan untuk 3 kategori
data_rows = []
for idx_b, b in enumerate(bulan_list):
    # Efek tren: penjualan sedikit naik menjelang pertengahan dan akhir tahun
    seasonal_factor = 1.0 + (0.15 * np.sin(idx_b / 11.0 * np.pi))

    for kat in kategori_list:
        if kat == 'Makanan & Minuman':
            base_trans = rng.randint(450, 700)
            avg_ticket = rng.uniform(45000, 65000)
        elif kat == 'Kebutuhan Rumah Tangga':
            base_trans = rng.randint(200, 380)
            avg_ticket = rng.uniform(85000, 130000)
        else: # Alat Tulis & Kantor
            base_trans = rng.randint(120, 250)
            avg_ticket = rng.uniform(25000, 45000)

        transaksi = int(base_trans * seasonal_factor)
        omzet = int(transaksi * avg_ticket)

        data_rows.append({
            'bulan_num': idx_b + 1,
            'bulan': b,
            'kategori': kat,
            'transaksi': transaksi,
            'penjualan': omzet
        })

df_sales = pd.DataFrame(data_rows)
df_sales['aov'] = (df_sales['penjualan'] / df_sales['transaksi']).round(0)

# Simpan ringkasan tahunan ke CSV
df_summary_tahunan = df_sales.groupby('kategori').agg({
    'penjualan': 'sum',
    'transaksi': 'sum'
}).reset_index()
df_summary_tahunan['aov_rata_rata'] = (df_summary_tahunan['penjualan'] / df_summary_tahunan['transaksi']).round(0)
df_summary_tahunan.to_csv("ringkasan_penjualan_tahunan.csv", index=False)

print(f"Data Penjualan Sintetis Berhasil Dibuat: {len(df_sales)} baris (12 bulan x 3 kategori).")
print("\nRingkasan Kinerja Penjualan Tahunan per Kategori:")
print(df_summary_tahunan.to_string(index=False))
```

### 6.2 Desain Awal (Sebelum Perbaikan - Desain Kurang Efektif)

Desain ini memperlihatkan contoh antarmuka yang kurang memedulikan kenyamanan pengguna:
- Tidak memiliki kartu ringkasan KPI (pengguna harus menghitung sendiri di kepala).
- Menggunakan warna acak yang saling bertabrakan dan garis kisi-kisi (*gridlines*) yang terlalu tebal.
- Skala angka pada sumbu tidak diformat (angka jutaan ditampilkan mentah seperti `40000000`).
- Ketiadaan hierarki judul yang terstruktur.

```python
# Desain Awal: Kurang Ergonomis
fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(14, 5), facecolor='#ffffcc') # Background kuning silau

# Plot 1: Garis tanpa hierarki dengan warna mencolok
colors_bad = ['#ff00ff', '#00ff00', '#00ffff']
for idx, kat in enumerate(kategori_list):
    sub = df_sales[df_sales['kategori'] == kat]
    ax1.plot(sub['bulan'], sub['penjualan'], marker='s', color=colors_bad[idx],
             linewidth=3, linestyle='--', label=kat)

ax1.set_title("GRAFIK OMZET SEMUA KATEGORI", fontsize=10, fontweight='bold', color='red')
ax1.set_ylabel("Penjualan dalam Rupiah mentah")
ax1.grid(True, which='both', color='black', linestyle='-', linewidth=1.5) # Grid terlalu mendominasi
ax1.legend(loc='lower left', fontsize=8)

# Plot 2: Bar chart bertumpuk tanpa format
pivot_df = df_sales.pivot(index='bulan', columns='kategori', values='transaksi')[kategori_list]
pivot_df.plot(kind='bar', stacked=True, ax=ax2, color=['magenta', 'lime', 'cyan'], edgecolor='black')
ax2.set_title("TOTAL TRANSAKSI", fontsize=10, color='blue')
ax2.set_ylabel("unit")
ax2.grid(True, color='gray', linestyle=':')

plt.suptitle("TAMPILAN AWAL DASHBOARD PENJUALAN (BELUM DIOPTIMASI)", fontsize=13, fontweight='bold', color='darkred')
plt.tight_layout()
plt.savefig("dashboard_penjualan_sebelum.png", dpi=300)
plt.show()
print("Grafik versi awal disimpan ke 'dashboard_penjualan_sebelum.png'.")
```

### 6.3 Desain Perbaikan (User-Centered Dashboard Berbasis Fungsi)

Menerapkan perbaikan komprehensif:
1. **Kartu KPI (*KPI Cards*)**: Menampilkan Omzet Total, Total Transaksi, dan Rata-rata Belanja (AOV) dengan format Rupiah yang ramah mata.
2. **Palet Warna Harmonis**: Menggunakan palet biru dan abu-abu korporat yang tenang dengan kontras rasio tinggi.
3. **Grafik Tren Garis**: Garis bersih dengan penanda titik data dan label sumbu dalam jutaan rupiah (Jt Rp).
4. **Grafik Batang Kategori**: Perbandingan total kategori dengan skala jujur yang dimulai dari titik nol ($0$).

```python
def render_user_centered_dashboard(kategori_pilihan="Semua Kategori"):
    # Filter data sesuai pilihan dropdown
    if kategori_pilihan == "Semua Kategori":
        df_filtered = df_sales.copy()
        label_filter = "Semua Kategori Produk"
    else:
        df_filtered = df_sales[df_sales['kategori'] == kategori_pilihan].copy()
        label_filter = f"Kategori: {kategori_pilihan}"

    # Hitung KPI
    total_omzet = df_filtered['penjualan'].sum()
    total_trans = df_filtered['transaksi'].sum()
    aov_rataan = total_omzet / total_trans if total_trans > 0 else 0

    # Format angka ramah manusia
    omzet_str = f"Rp {total_omzet/1e6:,.1f} Juta"
    trans_str = f"{total_trans:,} Transaksi"
    aov_str = f"Rp {aov_rataan:,.0f}"

    # Setup figure bersih modern
    fig = plt.figure(figsize=(14, 8), facecolor='#ffffff')
    gs = fig.add_gridspec(3, 2, height_ratios=[0.5, 1.2, 1.2], hspace=0.35, wspace=0.25)

    # 1. KPI Cards Header
    ax_kpi = fig.add_subplot(gs[0, :])
    ax_kpi.axis('off')

    # Kotak KPI 1: Omzet
    ax_kpi.text(0.16, 0.65, "TOTAL PENJUALAN (OMZET)", ha='center', va='center',
                fontsize=10, fontweight='bold', color='#7f8c8d')
    ax_kpi.text(0.16, 0.25, omzet_str, ha='center', va='center',
                fontsize=16, fontweight='bold', color='#2c3e50')

    # Kotak KPI 2: Transaksi
    ax_kpi.text(0.50, 0.65, "TOTAL TRANSAKSI", ha='center', va='center',
                fontsize=10, fontweight='bold', color='#7f8c8d')
    ax_kpi.text(0.50, 0.25, trans_str, ha='center', va='center',
                fontsize=16, fontweight='bold', color='#2980b9')

    # Kotak KPI 3: AOV
    ax_kpi.text(0.84, 0.65, "RATA-RATA TRANSAKSI (AOV)", ha='center', va='center',
                fontsize=10, fontweight='bold', color='#7f8c8d')
    ax_kpi.text(0.84, 0.25, aov_str, ha='center', va='center',
                fontsize=16, fontweight='bold', color='#27ae60')

    # Garis pembatas elegan
    ax_kpi.plot([0.02, 0.98], [0.0, 0.0], color='#bdc3c7', lw=1.2)

    # 2. Grafik Tren Bulanan (Kiri Bawah)
    ax_tren = fig.add_subplot(gs[1:, 0])
    monthly_trend = df_filtered.groupby(['bulan_num', 'bulan'])['penjualan'].sum().reset_index()

    ax_tren.plot(monthly_trend['bulan'], monthly_trend['penjualan'] / 1e6,
                 marker='o', color='#2980b9', linewidth=2.5, markersize=6, label='Penjualan')
    ax_tren.fill_between(monthly_trend['bulan'], monthly_trend['penjualan'] / 1e6,
                         color='#3498db', alpha=0.15)

    ax_tren.set_title(f"Tren Penjualan Bulanan\n({label_filter})", fontsize=11, fontweight='bold', pad=10)
    ax_tren.set_xlabel("Bulan", fontweight='bold')
    ax_tren.set_ylabel("Penjualan (Juta Rupiah)", fontweight='bold')
    ax_tren.set_ylim(bottom=0)
    ax_tren.grid(True, linestyle='--', alpha=0.5)

    # 3. Grafik Kontribusi per Kategori (Kanan Bawah)
    ax_bar = fig.add_subplot(gs[1:, 1])
    cat_summary = df_sales.groupby('kategori')['penjualan'].sum() / 1e6

    # Highlight kategori terpilih
    bar_colors = ['#2980b9' if (kategori_pilihan == "Semua Kategori" or k == kategori_pilihan) else '#bdc3c7'
                  for k in cat_summary.index]

    bars = ax_bar.barh(cat_summary.index, cat_summary.values, color=bar_colors, edgecolor='#34495e', height=0.55)
    ax_bar.set_title("Kontribusi Total Penjualan per Kategori (Tahunan)", fontsize=11, fontweight='bold', pad=10)
    ax_bar.set_xlabel("Total Penjualan (Juta Rupiah)", fontweight='bold')
    ax_bar.grid(True, linestyle='--', alpha=0.5, axis='x')

    # Label angka pada bar
    for b in bars:
        w = b.get_width()
        ax_bar.annotate(f"Rp {w:,.1f} Jt",
                        xy=(w, b.get_y() + b.get_height() / 2),
                        xytext=(6, 0), textcoords="offset points",
                        ha='left', va='center', fontsize=9, fontweight='bold')

    ax_bar.set_xlim(0, max(cat_summary.values) * 1.25)

    plt.suptitle("DASHBOARD PENJUALAN BERPUSAT PADA PENGGUNA (VERSI PERBAIKAN)",
                 fontsize=13, fontweight='bold', color='#2c3e50', y=0.98)

    plt.tight_layout()
    plt.savefig("dashboard_penjualan_sesudah.png", dpi=300)
    plt.show()

# Panggil fungsi untuk menampilkan versi default (Semua Kategori)
render_user_centered_dashboard("Semua Kategori")
print("Grafik versi perbaikan disimpan ke 'dashboard_penjualan_sesudah.png'.")
```

### 6.4 Implementasi Filter Dropdown Interaktif (Ipywidgets)

Filter dropdown berikut berfungsi interaktif di Google Colab. Pengguna dapat memilih kategori tertentu atau menampilkan seluruh kategori secara langsung tanpa memuat ulang notebook.

```python
# Widget Dropdown Kategori
dropdown_kategori = widgets.Dropdown(
    options=["Semua Kategori"] + kategori_list,
    value="Semua Kategori",
    description="Pilih Kategori:",
    style={'description_width': 'initial'},
    layout=widgets.Layout(width='320px')
)

output_dashboard = widgets.Output()

def on_category_change(change):
    if change['type'] == 'change' and change['name'] == 'value':
        with output_dashboard:
            clear_output(wait=True)
            render_user_centered_dashboard(change['new'])

dropdown_kategori.observe(on_category_change)

# Tampilkan widget dan output
display(dropdown_kategori)
with output_dashboard:
    render_user_centered_dashboard("Semua Kategori")
display(output_dashboard)
```
